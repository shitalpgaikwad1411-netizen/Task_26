```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

# Q1. (Data Preparation)

```python
#Load the dataset
df=pd.read_csv("cleaned_emotions.csv")
```

```python
# Separate the features (text) into X and target (emotion) into y.
X=df["cleaned_text"]
y=df["emotions"]

#Perform train-test split with test_size=0.20 and random_state=42.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

#Print the shape of X_train, X_test, y_train, and y_test.
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

# Q2. (Bag of Words – Basic)

```python
#Apply CountVectorizer (Bag of Words) on the training data:
from sklearn.feature_extraction.text import CountVectorizer

vectorizer=CountVectorizer()

#Fit and transform X_train.
X_train_bow=vectorizer.fit_transform(X_train)

#Transform X_test.
X_test_bow=vectorizer.transform(X_test)

#Print the shape of the resulting matrices.
print("X_train_bow shape:", X_train_bow.shape)
print("X_test_bow shape:", X_test_bow.shape)

#Display the first 20 feature names.
print("First 20 feature names:")
print(vectorizer.get_feature_names_out()[:20])
```

#Q3. (Bag of Words + MultinomialNB)

```python
#Train a Multinomial Naive Bayes model using the Bag of Words features:
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

model=MultinomialNB()

#Fit the model on X_train_bow and y_train.
model.fit(X_train_bow,y_train)

#Make predictions on the test set.
y_pred=model.predict(X_test_bow)

#Calculate and print the accuracy score.
accuracy=accuracy_score(y_test,y_pred)
print("Accuracy:", accuracy)
```

#Q4. (Understanding Vocabulary)

```python
#Print the total size of the vocabulary.
print("Total Size of Vocabulary:")
print(len(vectorizer.vocabulary_))

#Show any 15 words from the vocabulary.
print("15 words from the vocabulary:")
print(list(vectorizer.vocabulary_.keys())[:15])

#Convert one sample document from the training set into its BoW vector and display it.
sample_text=X_train.iloc[0]
sample_vector=vectorizer.transform([sample_text])

print("Sample text:")
print(sample_text)

print("BoW text:")
print(sample_vector.toarray())
```

#Q5. (N-grams with Bag of Words)

```python
#Create a new CountVectorizer that uses bigrams (ngram_range=(1, 2)).
bigram_vectorizer=CountVectorizer(ngram_range=(1,2))

#Fit and transform the training data.
X_train_bigram=bigram_vectorizer.fit_transform(X_train)

#Transform the test data.
X_test_bigram=bigram_vectorizer.transform(X_test)

#Print the shape of the new feature matrix.
print("X_train_bigram shape:", X_train_bigram.shape)
print("X_test_bigram shape:", X_test_bigram.shape)

#Display some of the bigram features.
print("First 20 feature names:")
print(bigram_vectorizer.get_feature_names_out()[:20])
```

#Q6. (N-grams + Model Training)

```python
#Train a MultinomialNB model using the bigram features created in Q5.
bigram_model=MultinomialNB()

bigram_model.fit(X_train_bigram,y_train)

y_pred_bigram=bigram_model.predict(X_test_bigram)

#Calculate and print the accuracy on the test set.
bigram_accuracy=accuracy_score(y_test,y_pred_bigram)
print("Bigram BoW Accuracy:", bigram_accuracy)

print("Unigram BoW Accuracy:", accuracy)
```

#Q7. (TF-IDF Vectorization)

```python
#Apply TfidfVectorizer on the same train-test split:
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf_vectorizer=TfidfVectorizer()

#Fit and transform X_train.
X_train_tfidf=tfidf_vectorizer.fit_transform(X_train)

#Transform X_test.
X_test_tfidf=tfidf_vectorizer.transform(X_test)

#Print the shape of the TF-IDF matrices.
print("X_train_tfidf shape:", X_train_tfidf.shape)
print("X_test_tfidf shape:", X_test_tfidf.shape)

#Display the first 15 feature names.
print("First 15 TF-IDF feature names:")
print(tfidf_vectorizer.get_feature_names_out()[:15])
```

#Q8. (TF-IDF + MultinomialNB)

```python
#Train a Multinomial Naive Bayes model using the TF-IDF features:
tfidf_model=MultinomialNB()

#Fit the model.
tfidf_model.fit(X_train_bigram,y_train)

#Make predictions on the test set.
y_pred_tfidf=tfidf_model.predict(X_test_bigram)

#Calculate and print the accuracy score.
tfidf_accuracy=accuracy_score(y_test,y_pred_tfidf)
print("TF-IDF Accuracy:", tfidf_accuracy)
```

#Q9. (Comparison of Vectorizers)

```python
comparison=pd.DataFrame({
    "Vectorizers": ["Bag of Words (Unigrams)", "Bag of Words (Unigrams + Bigrams)", "TF-IDF"],
    "Accuracy": [accuracy, bigram_accuracy, tfidf_accuracy]
})

print(comparison)
```
